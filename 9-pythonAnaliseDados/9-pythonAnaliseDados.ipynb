### **ARQUIVOS - python**

**TIPOS DE ARQUIVOS mais comuns**
**TXT** - texto simples
**CSV** - Tabelas e panilhas
**JSON** - Estruturas de dados, utilizado em APIS
**BANCO DE DADOS** - armanezamento estruturado SQL e NoSQL

**FUNCAO OPEN() -** serve para abrir o arquivo
open(”database.csv”)

**MODOS COMUNS DA FUNCAO**
**r -** para leitura
**w** - escrever no arquivo
**a** - comentario

```python
# CRIANDO UM ARQUIVO

with open("dados.txt", "w") as arquivo:
    arquivo.write(" Primeira linha - Python para analise de dados - ")
    arquivo.write(" Segunda linha - ABRINDO")

# LENDO O ARQUIVO
with open("dados.txt", "r") as arquivo:
    conteudo = arquivo.read()
    
print(conteudo)
print("-"*50)

# Lendo o arquivo linha por linha
with open("dados.txt", "r") as arquivo:
    for Linha in arquivo:
        print(Linha.strip())  # strip() para remover espaços em branco
```

```python
# escrevendo no arquivo
with open("relatorio.txt", "w") as arquivo1:
    arquivo1.write("Realtorio diario de vendas\n")
    arquivo1.write("Data: 29/11/2005\n")
    arquivo1.write("Total de vendas: R$ 1000,00\n")
    arquivo1.write("Vendedor: Leo\n")


# Adicionando mais uma linha no arquivo
with open("relatorio.txt", "a") as arquivo:
    arquivo.write("\nAdicionando mais uma linha no arquivo")
    
# lendo o arquivo
with open("relatorio.txt", "r") as arquivo1:
    for Linha in arquivo1:
        print(Linha.strip())
```

```python
# Leitura de Arquivos e escrita em CSV
import csv
dados = [
    ["nome","idade", "sobrenome"],
    ["Ana", 17, "Cadena"],
    ["Leo", 20, "Cadena"],
    ["Sophia", 17, "Cadena"]
]

# Utiliza o csv.writer - o writer do comeco e uma variavel que recebe qualquer nome, e o writerows escreve as linhas do arquivo csv, utilizando a variavel criada
with open("PessoasCSV.csv", "w", newline="") as arquivoCsv:
    writer = csv.writer(arquivoCsv)
    writer.writerows(dados)
    
# LENDO O ARQUIVO CSV
# Utiliza o csv.reader - o reader do comeco e uma variavel que recebe qualquer nome, e o for percorre cada linha do arquivo csv, imprimindo cada linha
with open("PessoasCSV.csv", "r") as arquivoCsv:
    reader = csv.reader(arquivoCsv)
    for Linha in reader:
        print(Linha)
```

```python
# JSON - utiliza o modulo json para ler e escrever arquivos json
import json

dados = {
    "nome": "ana",
    "idade": 17,
    "cidade": "São Paulo"
}

with open("dados.json", "w") as arquivoJson:
    json.dump(dados, arquivoJson)
    #json.dump() - recebe dois argumentos: o primeiro é o objeto Python que queremos escrever no arquivo JSON, e o segundo é o arquivo onde queremos escrever os dados. Ele converte o objeto Python em uma string JSON e escreve essa string no arquivo especificado.
    
# LENDO O ARQUIVO JSON
with open("dados.json", "r") as arquivosJson:
    dados_lidos = json.load(arquivosJson)
    # json.load() - recebe um argumento: o arquivo JSON que queremos ler. Ele lê o conteúdo do arquivo, converte a string JSON de volta para um objeto Python e retorna esse objeto.
    print(dados_lidos)
```

```python
# INTEGRANDO APIS
# - APi vai permitir que diferentes sistemas se conversem - buscando informacoes
# - Normalmente vai vir em json
# - Requisicoes HTTP - get, put, post, delete,
# - python utiliza a biblioteca request

import requests

url = "https://rickandmortyapi.com/api/character/2"
response = requests.get(url)
print(f"Status Code: {response.status_code}")  # Imprime o código de status da resposta
print(f"Response: {response.json()}")  # Imprime o conteúdo da resposta em formato JSON por conta do .json()


# imprime cada chave e valor do json
for chave, valor in response.json().items(): # utiliza o .items() para obter as chaves e valores do json, e o for percorre cada chave e valor, imprimindo cada um
    print(f"{chave}: {valor}")
```

```python

# BANCO DE DADOS - SQL - SQLite
''' INTEGRANDO COM O BANCO DE DADOS

- armazena em tabelas
- armazena grande quantidades de dados
- Consulta dados rapidamente 
- importar o sqllite3 para criar um banco de dados e manipular os dados
''' 

import sqlite3
conexao = sqlite3.connect("banco_de_dados.db") # cria um arquivo de banco de dados com o nome especificado
cursor = conexao.cursor() # cria um cursor para executar comandos SQL

cursor.execute('''CREATE TABLE IF NOT EXISTS usuarios
               (
                   nome TEXT,
                   idade INTEGER,
                     cidade TEXT
               )
               ''')

conexao.commit() # salva as alterações no banco de dados


# Inserindo dados na tabela
cursor.execute("INSERT INTO usuarios VALUES ('Leo', 20, 'SP')")
cursor.execute("INSERT INTO usuarios VALUES ('Ana', 17, 'SP')")


# lendo as informacoes
cursor.execute("SELECT * FROM usuarios") # seleciona todas as linhas da tabela usuarios
dados = cursor.fetchall() # fetchall() retorna todas as linhas selecionadas como uma lista de tuplas
print(dados)
```

```python
import sqlite3
import gc
import os

# conexão com context manager (fecha automaticamente)
with sqlite3.connect("banco_de_dados.db") as conexao:
    cursor = conexao.cursor()

    # cria tabela
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS usuarios (
            nome TEXT,
            idade INTEGER,
            cidade TEXT
        )
    ''')

    # inserção segura (parametrizada)
    usuarios = [
        ("Leo", 20, "SP"),
        ("Ana", 17, "SP")
    ]

    cursor.executemany(
        "INSERT INTO usuarios VALUES (?, ?, ?)", usuarios
    )

    conexao.commit()

    # leitura eficiente (evita fetchall em grandes volumes)
    cursor.execute("SELECT * FROM usuarios")

    for linha in cursor.fetchmany(10):  # lê em blocos
        print(linha)

# força limpeza de memória
gc.collect()


# ---------------------------
# LEITURA DE ARQUIVO 
# ---------------------------

arquivo = "dados.txt"

if os.path.exists(arquivo):
    try: 
        with open(arquivo, "r") as f:
            for linha in f:
                valor = linha.strip()

                print(valor)

                if valor.isdigit():
                    print(f"O valor {valor} é um número")
                else:
                    print(f"O valor {valor} não é um número")

    except Exception as e:
        print(f"Ocorreu um erro: {e}")
else:
    print(f"O arquivo {arquivo} não existe")
```

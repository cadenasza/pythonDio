TRATAMENTO E EXCECOES EM PYTHON
Tipos de Erros comuns 
  - SytaxeError       - Erro de sintaxe
  - ZeroDivisionError - Erro ao tentar dividir por 0
  - ValueError        - valor inválido
  - TypeError         - operação com tipo incorreto
  - FileNotFoundError - arquivo nao encontrado

```python
# Exemplo de erro
numero = 10 / 0  # resultará em um erro de divisão por zero
print(numero)
```

```python
# Tratando os errros
# TRY / EXCEPT / ELSE / FINALLY
# caso o usuario digite um número diferente de zero, o programa irá realizar a divisão e imprimir o resultado. Se o usuário digitar zero, o programa irá capturar a exceção ZeroDivisionError e imprimir uma mensagem de erro. O bloco else será executado apenas se não ocorrer nenhum erro, e o bloco finally será executado independentemente de ocorrer um erro ou não.
try:
    numero = int(input("Digite um número: "))
    resultado = 10 / numero
    print(f"O resultado da divisão é: {resultado}")
except ZeroDivisionError:
    print("Não é possível dividir por zero.")
else:
    print("A divisão foi realizada com sucesso.")
finally:
    print("Este bloco será executado independentemente de ocorrer um erro ou não.")
    
    
# Exemplo de tratamento de múltiplos erros
try:
    numero = int(input("Digite um número: "))
    resultado = 10 / numero
    print(f"O resultado da divisão é: {resultado}")
except ZeroDivisionError:
    print("Não é possível dividir por zero.")
except ValueError:
    print("Por favor, digite um número válido.")
    
    
# Exemplo de tratamento de erros genéricos
# Utilizando a classe base Exception, podemos capturar qualquer tipo de erro que possa ocorrer durante a execução do código. No exemplo abaixo, se o usuário digitar um valor que não seja um número ou se ocorrer uma divisão por zero, o programa irá capturar a exceção e imprimir uma mensagem de erro genérica.
try:
    numero = int(input("Digite um número: "))
    resultado = 10 / numero
    print(f"O resultado da divisão é: {resultado}")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
    
```

```python
# RAISE - Lançando exceções manualmente
# A palavra-chave raise é usada para lançar uma exceção manualmente. No exemplo abaixo, se o usuário digitar um número negativo, o programa irá lançar uma ValueError com uma mensagem personalizada.
try:
    numero = int(input("Digite um número: "))
    if numero < 0:
        raise ValueError("O número não pode ser negativo.")
    resultado = 10 / numero
    print(f"O resultado da divisão é: {resultado}")
except ValueError as e:
    print(f"Erro de valor: {e}")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
    
```

```python
# PREVENCAO DE ERROS - VALIDANDO ENTRADAS
# Validar as entradas do usuário é uma prática importante para prevenir erros e garantir que o programa funcione corretamente.

numero = input("Digite um número: ")
if numero.isdigit():
    numero = int(numero)
    print(f"O resultado é: {numero}")
else:
    print("Entrada inválida. Por favor, digite um número válido.")


# Mensagens claras de erros
def dividir(a, b):
    if b == 0:
        raise ValueError("O divisor não pode ser zero.")
    return a / b
try:
    resultado = dividir(10, 0)
    print(f"O resultado da divisão é: {resultado}")
except ValueError as e:
    print(f"Erro: {e}")
    
```
